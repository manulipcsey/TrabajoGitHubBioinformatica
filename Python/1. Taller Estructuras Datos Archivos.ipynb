```python
especies = {
    "Rattus": ["norvegicus", "rattus"],
    "Glaucidium": ["brasilianum", "hoskinsii", "californicum", "minutissimum"],
    "Echinocactus": ["grusonii"],
    "Opuntia": ["ficus-indica"],
}

claves = list(especies.keys())

especies["Panthera"] = ["leo", "tigris"]
especies["Quercus"] = ["robur"]

claves = list(especies.keys())   # se regenera para incluir los nuevos
print(set(claves))
```
